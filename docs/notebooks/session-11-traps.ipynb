### Trap 1 — predict, then run

```python
scores = [91, 58]   # a plain list, not an array
scores * 2
```

<details>
<summary><strong>What really happens — and why</strong></summary>

Result: `[91, 58, 91, 58]`

On a LIST, `* 2` repeats the list. Element-wise math is what NumPy arrays do: `np.array(scores) * 2` gives `array([182, 116])`.

</details>

### Trap 2 — predict, then run

```python
import numpy as np
scores = np.array([60, 75, 92])
(scores > 70) and (scores < 90)
```

<details>
<summary><strong>What really happens — and why</strong></summary>

Result: `ValueError: The truth value of an array with more than one element is ambiguous. Use a.any() or a.all()`

`and` needs ONE True/False, but each side is a whole array of them, so NumPy refuses to guess. Combine masks with `&` (and `|`, `~`), keeping the parentheses: `(scores > 70) & (scores < 90)`.

</details>

### Trap 3 — predict, then run

```python
import numpy as np
scores = np.array([91, 58, 73])
first_two = scores[:2]
first_two[0] = 0
scores
```

<details>
<summary><strong>What really happens — and why</strong></summary>

Result: `array([ 0, 58, 73])`

Slicing an ARRAY makes a view onto the same memory (slicing a list copies). Use `scores[:2].copy()` when you need an independent piece.

</details>

### Trap 4 — predict, then run

```python
import pandas as pd
likert = pd.Series([4, None, 5])   # one blank survey cell
likert.dtype
```

<details>
<summary><strong>What really happens — and why</strong></summary>

Result: `dtype('float64')`

Missing values are stored as NaN, and NaN is a float, so one blank turns the whole column into float64 (your 4 prints as 4.0).

</details>

### Trap 5 — predict, then run

```python
import pandas as pd
scores = pd.Series([91, 58, 73, 64])
scores.loc[0:2].tolist()
```

<details>
<summary><strong>What really happens — and why</strong></summary>

Result: `[91, 58, 73]`

`.loc` slices by LABEL and includes the end label. `.iloc[0:2]` slices by position and excludes the end, like normal Python.

</details>

### Trap 6 — predict, then run

```python
import pandas as pd
fall = pd.Series({'Ana': 85, 'Ben': 70})
spring = pd.Series({'Ben': 75, 'Cara': 90})
(spring - fall).to_dict()
```

<details>
<summary><strong>What really happens — and why</strong></summary>

Result: `{'Ana': nan, 'Ben': 5.0, 'Cara': nan}`

pandas lines Series up by their index LABELS, not by position. Labels present on only one side get NaN. Check with `.dropna()`, or fill first: `spring.sub(fall, fill_value=0)`.

</details>
