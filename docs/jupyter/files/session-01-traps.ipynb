### Trap 1 — predict, then run

```python
score = '5'
bonus = '3'   # what input() would hand you
score + bonus
```

<details>
<summary><strong>What really happens — and why</strong></summary>

Result: `'53'`

These are strings (as `input()` always returns): `+` concatenates text. Convert first: `int(score) + int(bonus)`.

</details>

### Trap 2 — predict, then run

```python
age = 21
'Age: ' + age
```

<details>
<summary><strong>What really happens — and why</strong></summary>

Result: `TypeError: can only concatenate str (not "int") to str`

You can't glue text to a number with `+`. Use an f-string — `f"Age: {age}"` — or `print('Age:', age)`.

</details>

### Trap 3 — predict, then run

```python
gpa = 3.9
int(gpa)
```

<details>
<summary><strong>What really happens — and why</strong></summary>

Result: `3`

`int()` truncates toward zero, it doesn't round. Use `round(gpa)` when you mean rounding.

</details>

### Trap 4 — predict, then run

```python
name = 'ada'
name.upper()
name
```

<details>
<summary><strong>What really happens — and why</strong></summary>

Result: `'ada'`

Strings are immutable: `.upper()` RETURNS a new string and leaves the original alone. Assign it back: `name = name.upper()`.

</details>
