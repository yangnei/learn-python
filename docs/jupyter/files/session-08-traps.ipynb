### Trap 1 — predict, then run

```python
import csv, io
row = next(csv.DictReader(io.StringIO('name,score\nAna,91')))
row['score'] + 1
```

<details>
<summary><strong>What really happens — and why</strong></summary>

Result: `TypeError: can only concatenate str (not "int") to str`

Every CSV value arrives as a STRING — `'91' + 1` doesn't add, it crashes. Convert first: `int(row['score']) + 1`.

</details>

### Trap 2 — predict, then run

```python
import json
json.dumps({'passed': True})
```

<details>
<summary><strong>What really happens — and why</strong></summary>

Result: `'{"passed": true}'`

JSON is its own language: Python's `True` is written as lowercase `true` (and `None` becomes `null`). `json.load` translates them back.

</details>

### Trap 3 — predict, then run

```python
import io
f = io.StringIO('Ana\nBen\n')   # stands in for an open file
first_pass = f.readlines()
f.readlines()
```

<details>
<summary><strong>What really happens — and why</strong></summary>

Result: `[]`

A file object is a cursor, not a container: after one pass it sits at the end and a second read returns nothing. Re-open the file (or `seek(0)`), or read it into a list once.

</details>
